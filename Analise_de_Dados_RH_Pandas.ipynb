## 1. Importação das bibliotecas e carregamento da base

Coloque `base_rh10.xlsx` na mesma pasta deste notebook. No Google Colab, também é possível fazer upload do arquivo quando solicitado.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

arquivo = Path("base_rh10.xlsx")

if not arquivo.exists():
    try:
        from google.colab import files
        enviados = files.upload()
        arquivo = Path(next(iter(enviados)))
    except ImportError:
        raise FileNotFoundError(
            "Não encontrei 'base_rh10.xlsx'. Coloque o arquivo na mesma pasta do notebook."
        )

rh = pd.read_excel(arquivo)
print(f"Dimensões da base: {rh.shape[0]} linhas e {rh.shape[1]} colunas")
rh.head()
```

## 2. Conhecendo os dados

## 3. Limpeza e preparação

Os campos categóricos ausentes são identificados como `Não informado`. Os campos numéricos são convertidos para formato numérico e os valores ausentes são preenchidos pela mediana. A coluna `ENTREVISTAS` é mantida para permitir análises.

```python
rh = rh.copy()

colunas_categoricas = ["CIDADE", "FONTE", "STATUS_FINAL"]
for coluna in colunas_categoricas:
    rh[coluna] = rh[coluna].fillna("Não informado")

for coluna in ["SCORE_TECNICO", "EXPECT_SALARIAL", "TEMPO_PROCESSO_DIAS"]:
    rh[coluna] = pd.to_numeric(rh[coluna], errors="coerce")
    if rh[coluna].notna().any():
        rh[coluna] = rh[coluna].fillna(rh[coluna].median())

duplicados = rh.duplicated().sum()
print(f"Registros duplicados encontrados: {duplicados}")
rh = rh.drop_duplicates().reset_index(drop=True)

print("\\nValores ausentes após tratamento:")
display(rh.isna().sum())
print(f"\\nDimensões após limpeza: {rh.shape[0]} linhas e {rh.shape[1]} colunas")
```

## 4. Candidaturas por vaga

```python
plt.figure(figsize=(10, 6))
rh["VAGA"].value_counts().sort_values().plot(kind="barh")
plt.xlabel("Quantidade de candidaturas")
plt.ylabel("Vaga")
plt.title("Candidaturas por vaga")
plt.tight_layout()
plt.show()
```

## 5. Distribuição dos status dos processos

```python
plt.figure(figsize=(8, 5))
rh["STATUS_FINAL"].value_counts().plot(kind="barh")
plt.xlabel("Quantidade de registros")
plt.ylabel("Status final")
plt.title("Distribuição dos status dos processos")
plt.tight_layout()
plt.show()
```

## 6. Fontes de recrutamento

```python
plt.figure(figsize=(9, 5))
rh["FONTE"].value_counts().plot(kind="bar")
plt.xlabel("Fonte de recrutamento")
plt.ylabel("Quantidade de candidaturas")
plt.title("Candidaturas por fonte")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()
plt.show()
```

## 7. Candidaturas por senioridade

```python
plt.figure(figsize=(8, 5))
rh["SENIORIDADE"].value_counts().plot(kind="bar")
plt.xlabel("Senioridade")
plt.ylabel("Quantidade de candidaturas")
plt.title("Candidaturas por senioridade")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. Indicadores numéricos

```python
colunas_numericas = ["SCORE_TECNICO", "EXPECT_SALARIAL", "TEMPO_PROCESSO_DIAS", "ENTREVISTAS"]
display(rh[colunas_numericas].describe().T)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for coluna, ax in zip(colunas_numericas, axes.flatten()):
    rh[coluna].dropna().plot(kind="hist", bins=20, ax=ax, title=f"Distribuição: {coluna}")
    ax.set_xlabel(coluna)
plt.tight_layout()
plt.show()
```
